## 4. Les listes

Une liste est une suite ordonnée de valeurs. Les colonnes d'un tableau, par exemple, forment une liste.

```python
# 4 — une liste, et ce qu'on peut lui demander
colonnes = ["clinique", "district", "sexe", "assurance"]

print("la liste entière   :", colonnes)
print("combien d'éléments :", len(colonnes))

print("le premier  :", colonnes[0], "  (on compte à partir de 0)")
print("le dernier  :", colonnes[-1])
print("du 2e au 3e :", colonnes[1:3], "  (le 4e est exclu)")

colonnes.append("quartier")   # ajouter à la fin
print("après l'ajout      :", colonnes)
```

## 5. Les dictionnaires

Un dictionnaire associe une **clé** à une **valeur**. Vous en verrez partout : chaque ligne d'un tableau d'essais, chaque constat chiffré, chaque prédiction écrite dans la soumission est un dictionnaire.

```python
# 5 — un dictionnaire : {clé: valeur}
patient = {"age": 41, "clinique": "A1", "absent": 0}

print("le dictionnaire entier :", patient)

print("l'âge, lu par sa clé :", patient["age"])
print("les clés            :", list(patient.keys()))
print("les valeurs         :", list(patient.values()))

patient["delai_jours"] = 12   # ajouter une clé
print("après l'ajout       :", patient)
```

## 6. Répéter : `for`

`for` parcourt une suite, élément par élément. C'est la boucle que vous verrez pour inspecter toutes les colonnes d'un tableau.

```python
# 6 — parcourir une liste
for c in colonnes[:3]:
    print("colonne :", c)

print("avec le rang \n")

# avec le rang, quand on en a besoin :
for rang, c in enumerate(colonnes[:3]):
    print("rang", rang, "—", c)

print("sur deux valeurs à la fois \n")

# sur deux valeurs à la fois (un tuple) — vous le verrez pour les fichiers :
for nom, extension in (("train", "csv"), ("test", "csv")):
    print("fichier :", nom + "." + extension)
```

## 7. Comprendre une suite en une ligne

Cette écriture — **la compréhension de liste** — est partout dans les notebooks du cours. Elle veut dire : « construis la liste des `c` tels que … ».

```python
# 7 — listes en compréhension
majuscules = [c.upper() for c in colonnes]
print("en majuscules :", majuscules)

# avec une condition — c'est LA forme que vous verrez le plus souvent
en_c = [c for c in colonnes if c.startswith("c")]
print("celles qui commencent par c :", en_c)

longueurs = [len(c) for c in colonnes]
print("leurs longueurs :", longueurs)

# et pour un dictionnaire, la même idée : {clé: valeur pour chaque élément}
rangs = {c: i for i, c in enumerate(colonnes)}
print("chaque colonne et son rang :", rangs)
```

## 8. Choisir : `if`, `else`, `continue`

`if` exécute un bloc seulement si la condition est vraie. `continue` passe directement à l'élément suivant de la boucle — vous le verrez pour **ignorer une colonne** qu'on ne sait pas encore traiter.

```python
# 8 — les conditions
age = 41

if age >= 40:
    print(age, "ans : 40 ans ou plus")
else:
    print(age, "ans : moins de 40 ans")

print("sauter une valeur dans la boucle \n")

# sauter un élément dans une boucle :
for c in colonnes:
    if c == "sexe":
        print("sautée  :", c)
        continue              # on ne traite pas cette colonne
    print("traitée :", c)

print("la forme courte \n")

# la forme courte, en une ligne :
etiquette = "long" if len(colonnes) > 10 else "court"
print("la liste est :", etiquette)
```

## 9. Mettre du code dans une fonction

Une fonction est un morceau de code qu'on nomme, pour le réutiliser. Dans ce cours, **tout ce que vous rendez doit pouvoir être appelé par quelqu'un d'autre** : c'est pour ça que vous écrirez des fonctions, et pas seulement des cellules qui s'enchaînent.

La ligne entre triples guillemets est une **docstring** : elle dit ce que la fonction fait.

```python
# 9 — définir, puis appeler
def moyenne(valeurs):
    """Renvoie la moyenne d'une liste de nombres."""
    return sum(valeurs) / len(valeurs)

print("moyenne de [10, 12, 14] :", moyenne([10, 12, 14]))

print("une fonction qui renvoie deux choses \n")

def resume(valeurs):
    """Renvoie DEUX choses à la fois."""
    return len(valeurs), max(valeurs)

# deux valeurs renvoyées se récupèrent sur une seule ligne :
combien, plus_grand = resume([10, 12, 14])
print("la liste compte", combien, "valeurs, la plus grande est", plus_grand)
```

## 10. Le tableau : `DataFrame`

C'est l'objet central du cours : un tableau, avec des colonnes nommées. Une colonne seule est une `Series` — c'est ce que renvoie `df["une_colonne"]`.

```python
# 10 — construire un petit tableau, et regarder dedans
df = pd.DataFrame({"age": [41, 55, 33],
                   "clinique": ["A1", "B2", "A1"],
                   "absent": [0, 1, 1]})

print("dimensions (lignes, colonnes) :", df.shape)
print("noms des colonnes            :", list(df.columns))

print("les trois lignes, ci-dessous :")
df.head()

print("la colonne age, seule :")
print(df["age"])

print("le type de chaque colonne :")
print(df.dtypes)
```

## 11. Calculer sur une colonne

Une colonne se comporte comme une suite de nombres : les mêmes opérations s'appliquent à toutes ses valeurs d'un coup, sans boucle.

```python
# 11 — quelques calculs
print("moyenne des âges :", df["age"].mean())
print("âge maximum      :", df["age"].max())

print("la colonne « absent » \n")

print("combien de 1 dans la colonne absent :", df["absent"].sum())
print("proportion de 1                     :", df["absent"].mean())
print("en pourcentage                      :", round(100 * df["absent"].mean(), 2), "%")

print("une comparaison, ligne par ligne \n")

# une comparaison renvoie True/False ligne par ligne, et un True vaut 1 :
print("plus de 40 ans, ligne par ligne :")
print(df["age"] > 40)
print("combien de lignes :", (df["age"] > 40).sum())
```

## 12. Garder certaines lignes

On filtre un tableau en lui donnant une condition entre crochets. C'est la forme que vous verrez pour regarder un sous-ensemble : les absents, les manquants, une clinique.

```python
# 12 — filtrer
absents = df[df["absent"] == 1]
print("les absents :", absents.shape, "(lignes, colonnes)")

print("âge moyen des absents :", df.loc[df["absent"] == 1, "age"].mean())

print("les lignes de plus de 40 ans, ci-dessous :")
df[df["age"] > 40]
```

## 13. Une liste de dictionnaires devient un tableau

C'est le geste que vous ferez pour **consigner vos essais** : on empile des dictionnaires, puis on en fait un tableau lisible. Le cours lira ce tableau, pas votre score.

```python
# 13 — consigner des essais
ESSAIS = []


def noter(hypothese, changement, score):
    """Ajoute une ligne au tableau des essais."""
    ESSAIS.append({"hypothèse": hypothese,
                   "changement": changement,
                   "score": round(score, 2)})
    print(pd.DataFrame(ESSAIS).to_string(index=False))


noter("rien faire", "classe majoritaire", 81.81)

print("un deuxième essai \n")

noter("un modèle suffit-il ?", "HistGradientBoosting", 82.77)
```

## 14. Mettre en forme un texte : les f-strings

Un `f` devant les guillemets permet d'insérer des valeurs dans le texte. Les nombres s'y arrondissent, et les colonnes s'y alignent — c'est ce qui rend un tableau imprimé lisible.

```python
# 14 — insérer des valeurs dans un texte
score = 82.76666

print(f"score arrondi : {score:.2f} %   (.2f = deux chiffres après la virgule)")

print("un tableau imprimé, aligné \n")

print("la longueur de chaque nom de colonne :")
for c in colonnes:
    print(f"  {c:<12} {len(c)} caractères   (<12 = aligné sur 12 caractères)")
```

## 16. Quand ça casse : lire le message

Une erreur n'est pas une punition : **elle dit exactement ce qui n'a pas marché**, et à quelle ligne. Deux messages que vous rencontrerez, et un `assert` — un test qu'on écrit soi-même pour vérifier que rien n'est cassé.

```python
# 16 — lire une erreur, et écrire son propre test
patient = {"age": 41}

try:
    print(patient["clinique"])
except KeyError as erreur:
    print("KeyError :", erreur)
    print("  → cette clé n'existe pas dans le dictionnaire")

print("un test qu'on écrit soi-même \n")

# un assert : la condition doit être vraie, sinon le message s'affiche
predictions = [0, 1, 1]
assert len(predictions) == 3, "attendu 3 prédictions"
print("test passé : 3 prédictions")

print("quand le test échoue \n")

try:
    assert len(predictions) == 500, f"attendu 500 prédictions, obtenu {len(predictions)}"
except AssertionError as erreur:
    print("un assert qui échoue dit, en clair, ce qui ne va pas :", erreur)
```
